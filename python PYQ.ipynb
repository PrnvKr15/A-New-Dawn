38. Write a recursive function to find the factorial of a number.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input("Enter a number = "))
print("Factorial of", n, "=", factorial(n))
```

39. Write a recursive function to find the sum of the first n natural numbers.

```python
def sum_natural(n):
    if n == 0:
        return 0
    return n + sum_natural(n - 1)

n = int(input("Enter n = "))
print("Sum of first", n, "natural numbers =", sum_natural(n))
```

49. Write a Python program using a named group in a regular expression to extract the year from a date.

```python
import re

date = input("Enter date (DD-MM-YYYY) = ")
match = re.search(r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})", date)
if match:
    print("Year =", match.group("year"))
else:
    print("Invalid date format")
```

60. Write a Python program to demonstrate polymorphism using two classes, Circle and Rectangle, each having an area method.

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

shapes = [Circle(7), Rectangle(10, 5)]
for shape in shapes:
    print(type(shape).__name__, "area =", round(shape.area(), 2))
```
